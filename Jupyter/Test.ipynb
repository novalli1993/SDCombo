```python
class Muli():
    def __init__(self):
        self.m = 1

    def __call__(self, plus):
        return self.m + plus

def main(args):
    m = Muli()
    num = m(2)
    return args ** num

if __name__ == '__main__':
    args = 3
    print(main(args))
```
